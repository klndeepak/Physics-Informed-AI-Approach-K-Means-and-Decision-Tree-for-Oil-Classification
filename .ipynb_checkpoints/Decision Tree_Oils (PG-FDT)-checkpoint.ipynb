# Physics-Guided Fingerprint Decision Tree (PG-FDT) - Oils

**The idea in one sentence:** instead of letting a decision tree pick from all ~1866 raw
wavenumber columns, we first build a small dictionary of Raman bands with a known chemical
meaning (validated against real reference spectra and published literature), and then only
let the tree choose from those.

**Why bother?** Two reasons. First, interpretability: every split the resulting tree makes
has a name and a chemical reason, not just a column index. Second, and more important,
this notebook directly **tests** whether that chemistry-based restriction actually helps -
or whether any block of about the same size, placed anywhere in the spectrum, would do just
as well. That comparison is the central result below, not an afterthought.

This notebook is self-contained: it reloads the data itself and does not depend on any other
notebook in this project.

```python
# Core data handling and plotting
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedGroupKFold, cross_val_score
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, confusion_matrix, balanced_accuracy_score, classification_report
from sklearn.base import clone  # used in Section 6 to fit a fresh tree per CV fold/trial
import seaborn as sns

RANDOM_STATE = 1  # one seed reused everywhere below, so the whole notebook is reproducible
```

## 1. Load the data

Same data-loading fix used throughout this project: spectral columns are sorted **ascending
by wavenumber** (not left in the source file's incidental order - this matters for tie-breaking
between highly correlated neighboring columns), and **no baseline shift** is applied, because
this file is already Z-score normalized upstream (a negative value here is a normal part of
that normalization, not an instrument artifact).

```python
def is_wavenumber(col):
    """A spectral column's header IS the wavenumber itself (e.g. "970.49"); metadata
    columns (the class label, the group id, ...) are not parseable as a float."""
    try:
        float(col); return True
    except ValueError:
        return False


df = pd.read_csv("Paper3_code\\Paper3_code\\data\\Raman_Oil_Baseline_corrected_smoothed_zscorenorm_filtered_500_1900_2600_3000_data_CS.csv")
df = df.drop(columns=[c for c in ['Unnamed: 0'] if c in df.columns])
spectral_cols = sorted([c for c in df.columns if is_wavenumber(c)], key=float)
meta_cols = [c for c in df.columns if c not in spectral_cols]
df = df[meta_cols + spectral_cols]

# x-axis, full raw spectral matrix, class label, and replicate-group id (used for
# group-aware CV in Section 5/6) - kept as four separate, clearly named arrays.
wavenumbers = np.array([float(w) for w in spectral_cols])
X = df[spectral_cols].astype(float)
y = df["Oil"]
groups = df["Oil_Type"]

X_train, X_test, y_train, y_test, groups_train, groups_test = train_test_split(
    X, y, groups, test_size=0.30, random_state=RANDOM_STATE, stratify=y)

print(f"{len(df)} spectra, {min(wavenumbers)}-{max(wavenumbers)} cm^-1, {len(spectral_cols)} wavenumber columns, "
      f"{y.nunique()} classes: {sorted(y.unique())}")
print(f"Train/test: {len(X_train)}/{len(X_test)}")
```

## 3. Cross-check against real reference spectra (`ramanbiolib`)

`ramanbiolib` supplies **reference spectra of pure compounds** - measured on a different
instrument, from the literature - not our oil samples. It is used for exactly one purpose
here: checking whether each dictionary window sits on a real, named vibrational feature in a
relevant pure fatty acid/triglyceride, including elaidic acid (the *trans* isomer of oleic
acid, produced by industrial hydrogenation and absent from natural oils). If a window is real
chemistry, it should land on a visible peak in these reference traces, not on flat baseline.

**Coverage limit, stated plainly:** every `ramanbiolib` spectrum used below is digitized only
over 450-1800 cm⁻¹ (checked directly against the package's own data). That covers 11 of our 14
windows. The three CH-stretch windows (`ch2_symmetric_stretch_2850`,
`ch2_asymmetric_stretch_2890`, `ch3_stretch_2935`) sit entirely outside that range and **cannot
be checked against `ramanbiolib` at all** - the cell below says so explicitly rather than
silently only plotting part of the dictionary and leaving the gap unexplained.

## 4. Extract fingerprint features

Two steps: `extract_windows` averages the Z-scored intensity inside each dictionary window
(one number per window, per spectrum), then `build_indices` turns pairs of those windows into
**ratios** (`safe_ratio`), not raw differences - e.g.
`cis_unsaturation_1655_1440 = cis_cc_stretch_1655 / |ch2_scissor_1440|`. Dividing by a broad,
always-present lipid reference band (mostly `ch2_scissor_1440`, the CH2 scissoring mode)
cancels out sample-to-sample differences in overall signal strength - laser power, focus,
acquisition gain - that a single window's raw intensity would otherwise carry, which a plain
difference does not. `safe_ratio` divides by the **absolute value** of the denominator (floored
at a tiny epsilon), so the ratio's sign always reflects the numerator, not an incidental sign
flip in a Z-scored reference band that happens to cross zero for a given spectrum.

```python
def extract_windows(X, wn, dictionary):
    """
    Extract the mean Raman intensity within each physics-guided window.

    Parameters
    ----------
    X : pd.DataFrame
        Spectra arranged as samples × wavenumber features.
    wn : array-like
        Wavenumber corresponding to each column of X.
    dictionary : dict
        Fingerprint definitions containing a ``window`` entry.

    Returns
    -------
    pd.DataFrame
        Mean intensity within every fingerprint window.
    """
    wn = np.asarray(wn, dtype=float)

    if X.shape[1] != len(wn):
        raise ValueError(f"X contains {X.shape[1]} columns, but wn contains {len(wn)} wavenumbers.")
    
    out = {}
    for name, spec in dictionary.items():
        lo, hi = spec["window"]
        mask = (wn >= lo) & (wn <= hi)

        if not np.any(mask):
            raise ValueError(
                f"Fingerprint '{name}' has no measured points inside [{lo:.1f}, {hi:.1f}] cm⁻¹.")

        out[name] = X.iloc[:, mask].mean(axis=1)

    return pd.DataFrame(out, index=X.index)


def safe_ratio(numerator, denominator, epsilon=1e-12):
    """
    Compute a numerically safe intensity ratio.
    """
    return numerator / denominator.abs().clip(lower=epsilon)


def build_indices(win):
    """
    Construct physically interpretable Raman indices.

    Ratios are preferred to raw differences because they reduce sensitivity
    to global intensity scaling, laser power, focusing and acquisition gain.
    """
    idx = pd.DataFrame(index=win.index)

    # Cis unsaturation relative to the broad lipid CH2 reference
    idx["cis_unsaturation_1655_1440"] = safe_ratio(win["cis_cc_stretch_1655"],win["ch2_scissor_1440"])
    # Independent cis-unsaturation marker
    idx["cis_bend_1265_1440"] = safe_ratio(win["cis_ch_bend_1265"],win["ch2_scissor_1440"])
    # Cis out-of-plane deformation relative to lipid abundance
    idx["cis_wag_970_1440"] = safe_ratio(win["cis_ch_wag_970"],win["ch2_scissor_1440"])
    # Relative trans-versus-cis double-bond contribution
    idx["trans_cis_1672_1655"] = safe_ratio(win["trans_cc_stretch_1672"],win["cis_cc_stretch_1655"])
    # Ordered-chain contribution relative to general lipid abundance
    idx["chain_order_1127_1440"] = safe_ratio(win["cc_skeletal_1127"], win["ch2_scissor_1440"])
    # Chain-order/conformation balance
    idx["chain_order_1127_1063"] = safe_ratio(win["cc_skeletal_1127"],win["ch2_rock_1063"])
    # Triglyceride ester contribution relative to lipid abundance
    idx["ester_1743_1440"] = safe_ratio(win["ester_carbonyl_1743"],win["ch2_scissor_1440"])
    # Saturated-chain CH2 contribution relative to terminal CH3
    idx["ch2_ch3_2850_2935"] = safe_ratio(win["ch2_symmetric_stretch_2850"],win["ch3_stretch_2935"])
    # Asymmetric-versus-symmetric CH2 stretching
    idx["ch2_asym_sym_2890_2850"] = safe_ratio(win["ch2_asymmetric_stretch_2890"],win["ch2_symmetric_stretch_2850"])

    return idx


# ----------------------------------------------------------------------
# Extract physics-guided features
# ----------------------------------------------------------------------

windows = extract_windows(X=X, wn=wavenumbers, dictionary=FINGERPRINT_DICTIONARY)
indices = build_indices(windows)
features = pd.concat([windows, indices], axis=1)

# Preserve the original train/test split
X_train_fp = features.loc[X_train.index].copy()
X_test_fp = features.loc[X_test.index].copy()

print(
    f"{windows.shape[1]} fingerprint-window features + "
    f"{indices.shape[1]} physics-derived indices = "
    f"{features.shape[1]} PG-FDT features, "
    f"versus {X.shape[1]} raw wavenumber features."
)
```

## 5. Two trees: full-spectrum baseline vs. PG-FDT

A CV-tuned baseline tree using **all** raw wavenumbers, and a PG-FDT tree using **only** the
fingerprint features above - both searched over the *same* `param_grid` and the *same*
`StratifiedGroupKFold` (grouped by `Oil_Type`, ~20 repeat spectra per group), so neither model
gets a hyperparameter-search advantage the other doesn't; the only thing that differs between
them is which features they are allowed to see. The comparison below reports accuracy, balanced
accuracy, weighted recall, tree size (depth/leaves), and the fraction of offered features each
tree actually used, plus a full per-class classification report for both models.

```python
# ----------------------------------------------------------------------
# Shared decision-tree search space
# ----------------------------------------------------------------------

param_grid = {
    "max_depth": [3, 4, 5],
    "max_leaf_nodes": [10, 16, 24],
    "min_samples_split": [10, 20],
}

# Use exactly the same grouped CV for both models
group_cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE,
)

base_tree = DecisionTreeClassifier(
    class_weight="balanced",
    random_state=RANDOM_STATE,
)

# ----------------------------------------------------------------------
# 1. Full-spectrum decision tree
# ----------------------------------------------------------------------

baseline_search = GridSearchCV(
    estimator=base_tree,
    param_grid=param_grid,
    scoring="recall_weighted",
    cv=group_cv,
    refit=True,
    n_jobs=-1,
    return_train_score=True,
)

baseline_search.fit(
    X_train,
    y_train,
    groups=groups_train,
)

baseline = baseline_search.best_estimator_

# ----------------------------------------------------------------------
# 2. Physics-guided fingerprint decision tree
# ----------------------------------------------------------------------

pgfdt_search = GridSearchCV(
    estimator=base_tree,
    param_grid=param_grid,
    scoring="recall_weighted",
    cv=group_cv,
    refit=True,
    n_jobs=-1,
    return_train_score=True,
)

pgfdt_search.fit(
    X_train_fp,
    y_train,
    groups=groups_train,
)

pgfdt = pgfdt_search.best_estimator_

# ----------------------------------------------------------------------
# Test-set predictions
# ----------------------------------------------------------------------

y_pred_baseline = baseline.predict(X_test)
y_pred_pgfdt = pgfdt.predict(X_test_fp)

# ----------------------------------------------------------------------
# Metrics
# ----------------------------------------------------------------------

results = pd.DataFrame(
    {
        "model": [
            "Full-spectrum DT",
            "Physics-guided FDT",
        ],
        "cv_weighted_recall": [
            baseline_search.best_score_,
            pgfdt_search.best_score_,
        ],
        "test_accuracy": [
            accuracy_score(y_test, y_pred_baseline),
            accuracy_score(y_test, y_pred_pgfdt),
        ],
        "test_balanced_accuracy": [
            balanced_accuracy_score(y_test, y_pred_baseline),
            balanced_accuracy_score(y_test, y_pred_pgfdt),
        ],
        "test_weighted_recall": [
            recall_score(
                y_test,
                y_pred_baseline,
                average="weighted",
                zero_division=0,
            ),
            recall_score(
                y_test,
                y_pred_pgfdt,
                average="weighted",
                zero_division=0,
            ),
        ],
        "selected_features": [
            int(np.count_nonzero(baseline.feature_importances_)),
            int(np.count_nonzero(pgfdt.feature_importances_)),
        ],
        "available_features": [
            X_train.shape[1],
            X_train_fp.shape[1],
        ],
        "tree_depth": [
            baseline.get_depth(),
            pgfdt.get_depth(),
        ],
        "leaf_count": [
            baseline.get_n_leaves(),
            pgfdt.get_n_leaves(),
        ],
    }
)

results["feature_fraction"] = (
    results["selected_features"] / results["available_features"]
)

print("\nModel comparison")
print(results.round(4).to_string(index=False))

print("\nBest full-spectrum parameters:")
print(baseline_search.best_params_)

print("\nBest PG-FDT parameters:")
print(pgfdt_search.best_params_)

# ----------------------------------------------------------------------
# Detailed test reports
# ----------------------------------------------------------------------

print("\nFull-spectrum decision-tree report:")
print(
    classification_report(
        y_test,
        y_pred_baseline,
        digits=3,
        zero_division=0,
    )
)

print("\nPG-FDT classification report:")
print(
    classification_report(
        y_test,
        y_pred_pgfdt,
        digits=3,
        zero_division=0,
    )
)

# ----------------------------------------------------------------------
# PG-FDT rules
# ----------------------------------------------------------------------

print("\nPG-FDT decision rules:")
print(
    export_text(
        pgfdt,
        feature_names=list(X_train_fp.columns),
        show_weights=True,
        decimals=4,
    )
)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

labels = ['GNO', 'PO', 'SO', 'SOYO', 'VO']

y_pred = pgfdt.predict(X_test_fp)

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=labels,
    normalize=None
)

fig, ax = plt.subplots(figsize=(5, 4))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)

disp.plot(
    cmap="Blues",
    values_format="d",
    colorbar=False,
    ax=ax
)

ax.set_title("PG-FDT Test Confusion Matrix")
plt.tight_layout()
plt.savefig("pgfdt_confusion_matrix_oils.jpg", dpi=300, bbox_inches="tight")
plt.show()
```

## 6. Is this actually about chemistry, or would any same-size block do?

This is the key question: window-averaging alone reduces per-column noise regardless of *where*
a window sits, so a good result could just mean "averaging helps", not "these specific bands
matter". Three conditions, evaluated on the *exact same* frozen `StratifiedGroupKFold` folds
(built once and reused everywhere below), so none of the comparison is confounded by different
fold splits:

- **Physics-guided dictionary** - our 14 real windows, variable widths, as chosen from
  chemistry.
- **Unrestricted fixed-width tiling** - the *whole* spectrum chopped into non-overlapping
  blocks at a single width (the median width across our dictionary windows), with zero chemical
  knowledge, then a tree picks freely among all of them.
- **Random equal-budget layouts** - 14 windows at random positions, matching our dictionary's
  widths one-for-one, repeated many times to build a null distribution.

If arbitrary blocks did just as well, the "physics-guided" framing would be decoration. The
report below gives where the dictionary's score falls in that random-layout distribution (a
percentile and a one-sided Monte Carlo p-value), and - just as tellingly - which wavenumbers
the *unrestricted* tiling search chose on its own, so we can check whether it independently
rediscovers regions our dictionary already flagged.

```python
# ---------------------------------------------------------------------
# Valid measured Raman regions
# ---------------------------------------------------------------------

REGIONS = [
    (500.8563, 1900.0),
    (2600.0, 2999.186),
]


# ---------------------------------------------------------------------
# Feature extraction
# ---------------------------------------------------------------------

def mean_in_window(Xv, wn, lo, hi):
    """Mean intensity for each spectrum inside [lo, hi]."""
    wn = np.asarray(wn, dtype=float)
    mask = (wn >= lo) & (wn <= hi)

    if not np.any(mask):
        raise ValueError(
            f"No measured Raman points in window "
            f"[{lo:.2f}, {hi:.2f}] cm⁻¹."
        )

    return np.nanmean(Xv[:, mask], axis=1)


def extract_window_features(Xv, wn, windows):
    """Create one mean-intensity feature per Raman window."""
    return np.column_stack([
        mean_in_window(Xv, wn, lo, hi)
        for lo, hi in windows
    ])
```

```python
# ---------------------------------------------------------------------
# Fixed tiling baseline
# ---------------------------------------------------------------------

def tile_windows(width, regions=REGIONS):
    """Generate adjacent, non-overlapping windows."""
    windows = []

    for region_lo, region_hi in regions:
        start = region_lo

        while start + width <= region_hi:
            windows.append((start, start + width))
            start += width

    return windows


# ---------------------------------------------------------------------
# Random equal-budget baseline
# ---------------------------------------------------------------------

def random_window(rng, width, regions=REGIONS):
    """
    Draw one window uniformly over all valid starting positions.

    Regions are weighted by the amount of space available for a window
    of the requested width.
    """
    available_lengths = np.array([
        max(0.0, region_hi - region_lo - width)
        for region_lo, region_hi in regions
    ])

    valid_regions = np.flatnonzero(available_lengths > 0)

    if len(valid_regions) == 0:
        raise ValueError(
            f"Window width {width:.2f} cm⁻¹ does not fit "
            "inside any valid Raman region."
        )

    probabilities = available_lengths[valid_regions]
    probabilities = probabilities / probabilities.sum()

    region_index = rng.choice(
        valid_regions,
        p=probabilities,
    )

    region_lo, region_hi = regions[region_index]

    lo = rng.uniform(
        region_lo,
        region_hi - width,
    )

    return lo, lo + width


def random_layout(rng, widths, regions=REGIONS):
    """Generate one random layout preserving dictionary window widths."""
    return [
        random_window(rng, width, regions)
        for width in widths
    ]
```

```python
# ---------------------------------------------------------------------
# Explicit grouped-CV evaluator
# ---------------------------------------------------------------------

def grouped_cv_scores(
    estimator,
    features,
    y,
    groups,
    splits,
):
    """
    Evaluate balanced accuracy using precomputed identical grouped folds.
    """
    y = np.asarray(y)
    scores = []

    for train_idx, valid_idx in splits:
        model = clone(estimator)

        model.fit(
            features[train_idx],
            y[train_idx],
        )

        prediction = model.predict(
            features[valid_idx]
        )

        scores.append(
            balanced_accuracy_score(
                y[valid_idx],
                prediction,
            )
        )

    return np.asarray(scores)


# ---------------------------------------------------------------------
# Shared model and grouped folds
# ---------------------------------------------------------------------

fixed_tree = DecisionTreeClassifier(
    class_weight="balanced",
    random_state=RANDOM_STATE,
    max_depth=4,
    max_leaf_nodes=16,
    min_samples_split=10,
)

group_cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE,
)

Xv_train = X_train.to_numpy(dtype=float)
y_train_array = np.asarray(y_train)
groups_train_array = np.asarray(groups_train)

# Freeze the fold assignments so every method sees exactly the same folds.
cv_splits = list(
    group_cv.split(
        Xv_train,
        y_train_array,
        groups_train_array,
    )
)
```

```python
# =====================================================================
# Condition 1: Physics-guided chemistry dictionary
# =====================================================================

dict_names = list(FINGERPRINT_DICTIONARY.keys())

dict_windows = [
    tuple(spec["window"])
    for spec in FINGERPRINT_DICTIONARY.values()
]

dict_widths = np.array([
    hi - lo
    for lo, hi in dict_windows
])

feats_dict = extract_window_features(
    Xv_train,
    wavenumbers,
    dict_windows,
)

cv_dict = grouped_cv_scores(
    estimator=fixed_tree,
    features=feats_dict,
    y=y_train_array,
    groups=groups_train_array,
    splits=cv_splits,
)
```

```python
# =====================================================================
# Condition 2: Unrestricted fixed tiling
# =====================================================================

# The fixed window width for the unrestricted baseline was set to 26 cm⁻¹, 
# corresponding to the mean (26.1 cm⁻¹) and median (25.5 cm⁻¹) widths of the physics-guided fingerprint windows.
tile_width = float(np.median(dict_widths))

tiles = tile_windows(
    width=tile_width,
    regions=REGIONS,
)

feats_tiles = extract_window_features(
    Xv_train,
    wavenumbers,
    tiles,
)

cv_tiles = grouped_cv_scores(
    estimator=fixed_tree,
    features=feats_tiles,
    y=y_train_array,
    groups=groups_train_array,
    splits=cv_splits,
)

# Fit once on all training data only to inspect which tiles are selected.
tile_tree = clone(fixed_tree)
tile_tree.fit(feats_tiles, y_train_array)

used_tiles = np.flatnonzero(
    tile_tree.feature_importances_ > 0
)


def overlap_length(window_a, window_b):
    """Intersection length of two Raman windows."""
    alo, ahi = window_a
    blo, bhi = window_b

    return max(
        0.0,
        min(ahi, bhi) - max(alo, blo),
    )


def overlaps_dictionary(tile, dictionary_windows, minimum_fraction=0.0):
    """
    Determine whether a tile overlaps a chemistry window.

    minimum_fraction is the required fraction of the tile covered by
    at least one dictionary window.
    """
    tile_width_local = tile[1] - tile[0]

    return any(
        overlap_length(tile, dictionary_window)
        / tile_width_local > minimum_fraction
        for dictionary_window in dictionary_windows
    )


# Requiring 25% overlap is more meaningful than counting boundary contact.
minimum_overlap_fraction = 0.25

overlap_flags = [
    overlaps_dictionary(
        tiles[i],
        dict_windows,
        minimum_fraction=minimum_overlap_fraction,
    )
    for i in used_tiles
]

overlap_count = int(np.sum(overlap_flags))
```

```python
# =====================================================================
# Condition 3: Random same-number, same-width layouts
# =====================================================================

rng = np.random.default_rng(RANDOM_STATE)

n_trials = 5000
random_scores = np.empty(n_trials, dtype=float)

for trial in range(n_trials):
    layout = random_layout(
        rng=rng,
        widths=dict_widths,
        regions=REGIONS,
    )

    random_features = extract_window_features(
        Xv_train,
        wavenumbers,
        layout,
    )

    fold_scores = grouped_cv_scores(
        estimator=fixed_tree,
        features=random_features,
        y=y_train_array,
        groups=groups_train_array,
        splits=cv_splits,
    )

    random_scores[trial] = fold_scores.mean()
```

```python
import numpy as np
import pandas as pd


# =====================================================================
# 1. Validate results from the three experimental conditions
# =====================================================================

cv_dict = np.asarray(cv_dict, dtype=float)
cv_tiles = np.asarray(cv_tiles, dtype=float)
random_scores = np.asarray(random_scores, dtype=float)

if cv_dict.size == 0:
    raise ValueError("cv_dict is empty.")

if cv_tiles.size == 0:
    raise ValueError("cv_tiles is empty.")

if random_scores.size == 0:
    raise ValueError("random_scores is empty.")

if not np.all(np.isfinite(cv_dict)):
    raise ValueError("cv_dict contains NaN or infinite values.")

if not np.all(np.isfinite(cv_tiles)):
    raise ValueError("cv_tiles contains NaN or infinite values.")

if not np.all(np.isfinite(random_scores)):
    raise ValueError("random_scores contains NaN or infinite values.")

n_trials_actual = random_scores.size

if n_trials_actual != n_trials:
    print(
        f"Warning: n_trials={n_trials}, but random_scores contains "
        f"{n_trials_actual} completed trials."
    )


# =====================================================================
# 2. Summarize grouped cross-validation performance
# =====================================================================

dictionary_mean = cv_dict.mean()
dictionary_std = cv_dict.std(ddof=1)

tile_mean = cv_tiles.mean()
tile_std = cv_tiles.std(ddof=1)

random_mean = random_scores.mean()
random_std = random_scores.std(ddof=1)
random_median = np.median(random_scores)

random_ci_low, random_ci_high = np.percentile(
    random_scores,
    [2.5, 97.5],
)

random_best = random_scores.max()
random_worst = random_scores.min()


# =====================================================================
# 3. Compare the chemistry dictionary with random layouts
# =====================================================================

n_random_worse = np.sum(random_scores < dictionary_mean)
n_random_equal_or_better = np.sum(
    random_scores >= dictionary_mean
)

# Percentage of random layouts performing worse than the dictionary.
dictionary_percentile = (
    100.0 * n_random_worse / n_trials_actual
)

# One-sided Monte Carlo p-value:
# H0: random placement performs at least as well as chemistry guidance.
empirical_p = (
    1 + n_random_equal_or_better
) / (
    n_trials_actual + 1
)

dictionary_minus_random = dictionary_mean - random_mean

# Standardized difference relative to random-layout variability.
if random_std > 0:
    standardized_advantage = (
        dictionary_minus_random / random_std
    )
else:
    standardized_advantage = np.nan


# =====================================================================
# 4. Summarize unrestricted-tile selection
# =====================================================================

used_tiles = np.asarray(used_tiles, dtype=int)
overlap_flags = np.asarray(overlap_flags, dtype=bool)

if len(used_tiles) != len(overlap_flags):
    raise ValueError(
        "used_tiles and overlap_flags must have the same length."
    )

n_used_tiles = len(used_tiles)
n_overlapping_tiles = int(overlap_flags.sum())

if n_used_tiles > 0:
    overlap_percentage = (
        100.0 * n_overlapping_tiles / n_used_tiles
    )
else:
    overlap_percentage = np.nan


selected_tile_rows = []

for tile_index, overlap_flag in zip(
    used_tiles,
    overlap_flags,
):
    lo, hi = tiles[tile_index]

    selected_tile_rows.append(
        {
            "tile_index": int(tile_index),
            "window_lo_cm-1": float(lo),
            "window_hi_cm-1": float(hi),
            "importance": float(
                tile_tree.feature_importances_[tile_index]
            ),
            "dictionary_associated": bool(overlap_flag),
        }
    )

selected_tiles_table = pd.DataFrame(selected_tile_rows)

if not selected_tiles_table.empty:
    selected_tiles_table = selected_tiles_table.sort_values(
        "importance",
        ascending=False,
    ).reset_index(drop=True)


# =====================================================================
# 5. Create a compact numerical summary table
# =====================================================================

summary_table = pd.DataFrame(
    {
        "condition": [
            "Physics-guided dictionary",
            "Unrestricted fixed-width tiling",
            "Random equal-budget layouts",
        ],
        "candidate_windows": [
            len(dict_windows),
            len(tiles),
            len(dict_windows),
        ],
        "window_strategy": [
            "Chemistry-guided, variable width",
            f"Complete non-overlapping tiling, {tile_width:.1f} cm^-1",
            "Random positions, dictionary-matched widths",
        ],
        "mean_balanced_accuracy": [
            dictionary_mean,
            tile_mean,
            random_mean,
        ],
        "std_balanced_accuracy": [
            dictionary_std,
            tile_std,
            random_std,
        ],
    }
)


# =====================================================================
# 6. Print the complete report
# =====================================================================

separator = "=" * 76

print(separator)
print("PG-FDT WINDOW-SELECTION COMPARISON")
print(separator)

print("\n1. PERFORMANCE SUMMARY\n")

print(
    summary_table.to_string(
        index=False,
        formatters={
            "mean_balanced_accuracy": "{:.4f}".format,
            "std_balanced_accuracy": "{:.4f}".format,
        },
    )
)

print("\n2. PHYSICS-GUIDED DICTIONARY\n")

print(
    f"Number of fingerprint windows : {len(dict_windows)}\n"
    f"Fold scores                  : "
    f"{np.round(cv_dict, 4).tolist()}\n"
    f"Mean balanced accuracy       : {dictionary_mean:.4f}\n"
    f"Fold-to-fold SD              : {dictionary_std:.4f}"
)

print("\n3. UNRESTRICTED TILING\n")

print(
    f"Fixed tile width             : {tile_width:.2f} cm^-1\n"
    f"Candidate tiles              : {len(tiles)}\n"
    f"Fold scores                  : "
    f"{np.round(cv_tiles, 4).tolist()}\n"
    f"Mean balanced accuracy       : {tile_mean:.4f}\n"
    f"Fold-to-fold SD              : {tile_std:.4f}\n"
    f"Tiles used by fitted tree    : {n_used_tiles}\n"
    f"Dictionary-associated tiles  : "
    f"{n_overlapping_tiles}/{n_used_tiles}"
)

if n_used_tiles > 0:
    print(
        f"Overlap percentage           : "
        f"{overlap_percentage:.1f}%\n"
        f"Minimum overlap criterion    : "
        f"{100 * minimum_overlap_fraction:.0f}% of tile width"
    )

    print("\nSelected unrestricted tiles:\n")

    print(
        selected_tiles_table.to_string(
            index=False,
            formatters={
                "window_lo_cm-1": "{:.1f}".format,
                "window_hi_cm-1": "{:.1f}".format,
                "importance": "{:.4f}".format,
            },
        )
    )
else:
    print("The fitted unrestricted tree did not use any tiles.")

print("\n4. RANDOM EQUAL-BUDGET BASELINE\n")

print(
    f"Completed random trials      : {n_trials_actual}\n"
    f"Windows per layout           : {len(dict_windows)}\n"
    f"Mean balanced accuracy       : {random_mean:.4f}\n"
    f"Median balanced accuracy     : {random_median:.4f}\n"
    f"Random-layout SD             : {random_std:.4f}\n"
    f"95% layout interval          : "
    f"[{random_ci_low:.4f}, {random_ci_high:.4f}]\n"
    f"Worst random layout          : {random_worst:.4f}\n"
    f"Best random layout           : {random_best:.4f}"
)

print("\n5. DICTIONARY VERSUS RANDOM PLACEMENT\n")

print(
    f"Dictionary score             : {dictionary_mean:.4f}\n"
    f"Random-layout mean           : {random_mean:.4f}\n"
    f"Absolute advantage           : "
    f"{dictionary_minus_random:+.4f}\n"
    f"Standardized advantage       : "
    f"{standardized_advantage:+.2f} random SD\n"
    f"Random layouts worse         : "
    f"{n_random_worse}/{n_trials_actual}\n"
    f"Random layouts equal/better  : "
    f"{n_random_equal_or_better}/{n_trials_actual}\n"
    f"Dictionary percentile        : "
    f"{dictionary_percentile:.2f}%\n"
    f"Empirical one-sided p-value  : {empirical_p:.6f}"
)

print("\n6. AUTOMATIC INTERPRETATION\n")

if empirical_p < 0.05:
    significance_text = (
        "The physics-guided dictionary performs significantly better "
        "than random equal-budget window placement."
    )
else:
    significance_text = (
        "The physics-guided dictionary is not significantly better "
        "than random equal-budget window placement at alpha = 0.05."
    )

if dictionary_mean > random_mean:
    direction_text = (
        "Its mean grouped-CV balanced accuracy is higher than the "
        "random-layout mean."
    )
elif dictionary_mean < random_mean:
    direction_text = (
        "Its mean grouped-CV balanced accuracy is lower than the "
        "random-layout mean."
    )
else:
    direction_text = (
        "Its mean grouped-CV balanced accuracy equals the "
        "random-layout mean."
    )

if dictionary_percentile >= 95:
    percentile_text = (
        "The dictionary lies within the top 5% of tested "
        "equal-budget window layouts."
    )
elif dictionary_percentile >= 90:
    percentile_text = (
        "The dictionary lies within the top 10% of tested "
        "equal-budget window layouts."
    )
elif dictionary_percentile >= 50:
    percentile_text = (
        "The dictionary performs above the median random layout, "
        "but not within the top 10%."
    )
else:
    percentile_text = (
        "At least half of the random layouts outperform the dictionary."
    )

print(significance_text)
print(direction_text)
print(percentile_text)

print(separator)
```

## 7. Uncertainty: let the tree say "not sure"

Instead of always forcing an answer, we can withhold a prediction when the tree's confidence
(training-class-proportion at the landed leaf) is low - trading coverage for accuracy on the
cases we do answer. The table below sweeps the confidence threshold from 0 to 0.95 and reports
coverage alongside four accuracy metrics; the two plots after it show the same trade-off
visually, plus where prediction confidence tends to sit across the whole test set.

```python
# For each confidence threshold: "kept"/"coverage" = how many test samples the tree is
# willing to answer (confidence >= threshold); the four metric columns are computed only
# on those kept samples.
confidence = pgfdt.predict_proba(X_test_fp).max(axis=1)
predictions = pgfdt.predict(X_test_fp)

thresholds = np.arange(0.0, 1.0, 0.05)

rows = []

for t in thresholds:

    keep = confidence >= t

    n_keep = keep.sum()
    n_total = len(keep)

    if n_keep == 0:
        rows.append({
            "threshold": t,
            "kept": 0,
            "rejected": n_total,
            "coverage (%)": 0,
            "accuracy": np.nan,
            "balanced_accuracy": np.nan,
            "weighted_recall": np.nan,
            "macro_recall": np.nan,
        })
        continue

    y_true = y_test[keep]
    y_pred = predictions[keep]

    rows.append({
        "threshold": t,
        "kept": n_keep,
        "rejected": n_total - n_keep,
        "coverage (%)": 100 * n_keep / n_total,
        "accuracy": accuracy_score(y_true, y_pred),
        "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
        "weighted_recall": recall_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0,
        ),
        "macro_recall": recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0,
        ),
    })

confidence_table = pd.DataFrame(rows)

print(confidence_table.round(3))

confidence_table.to_csv(
    "pgfdt_confidence_thresholds.csv",
    index=False,
)
```

## 8. Examples, explained

Because every feature has a name, `explain_tree_prediction` turns the tree's root-to-leaf
decision path for any one sample into a short chemical argument instead of a list of column
thresholds - including the full class-probability distribution at the landed leaf, not just the
winning class. The first pair of examples below is drawn at random; the second pair is chosen
deliberately - one correct prediction and one incorrect one - since a single random draw isn't
guaranteed to include a mistake worth looking at.

```python
def explain_tree_prediction(
    model,
    row,
    feature_names=None,
    true_label=None,
    show_leaf_distribution=True,
    decimals=3,
):
    """
    Explain one decision-tree prediction by printing the complete
    root-to-leaf decision path.

    Parameters
    ----------
    model : sklearn.tree.DecisionTreeClassifier
        Fitted decision-tree classifier.

    row : pd.Series
        One sample whose index contains the feature names.

    feature_names : list-like, optional
        Feature names in the same order used to train the model.
        If None, row.index is used.

    true_label : optional
        Known class label for this sample.

    show_leaf_distribution : bool
        Whether to print the class distribution in the terminal leaf.

    decimals : int
        Number of decimal places shown for values and thresholds.
    """
    if not isinstance(row, pd.Series):
        raise TypeError("row must be a pandas Series.")

    if feature_names is None:
        feature_names = list(row.index)
    else:
        feature_names = list(feature_names)

    if len(feature_names) != model.n_features_in_:
        raise ValueError(
            f"Model expects {model.n_features_in_} features, "
            f"but {len(feature_names)} names were supplied."
        )

    missing = [name for name in feature_names if name not in row.index]
    if missing:
        raise ValueError(
            f"The row is missing these model features: {missing}"
        )

    # Preserve exactly the feature order used during model fitting.
    sample = row.loc[feature_names].to_frame().T

    probabilities = model.predict_proba(sample)[0]
    predicted_index = int(np.argmax(probabilities))
    predicted_label = model.classes_[predicted_index]
    predicted_probability = probabilities[predicted_index]

    # Leaf probability is based on the class composition of the leaf;
    # it should not automatically be interpreted as calibrated confidence.
    correctness = ""
    if true_label is not None:
        status = "CORRECT" if predicted_label == true_label else "INCORRECT"
        correctness = f" | True: {true_label} | {status}"

    print("=" * 76)
    print(
        f"Predicted: {predicted_label}"
        f" | Leaf probability: {predicted_probability:.{decimals}f}"
        f"{correctness}"
    )

    probability_text = ", ".join(
        f"{label}={probability:.{decimals}f}"
        for label, probability in zip(model.classes_, probabilities)
    )
    print(f"Class probabilities: {probability_text}")

    decision_path = model.decision_path(sample)
    node_ids = decision_path.indices
    leaf_id = int(model.apply(sample)[0])

    print("\nDecision path:")

    step = 1

    for node_id in node_ids:
        feature_index = model.tree_.feature[node_id]

        # Negative feature index means that this is a leaf.
        if feature_index < 0:
            continue

        feature_name = feature_names[feature_index]
        threshold = model.tree_.threshold[node_id]
        value = float(sample.iloc[0, feature_index])

        went_left = value <= threshold
        operator = "<=" if went_left else ">"
        direction = "left" if went_left else "right"

        print(
            f"  Step {step}: {feature_name}\n"
            f"           observed value = {value:.{decimals}f}\n"
            f"           rule           = value {operator} "
            f"{threshold:.{decimals}f}\n"
            f"           branch taken   = {direction}"
        )

        step += 1

    print(f"\nTerminal leaf: node {leaf_id}")

    if show_leaf_distribution:
        leaf_counts = model.tree_.value[leaf_id][0]
        leaf_total = leaf_counts.sum()

        if leaf_total > 0:
            leaf_distribution = leaf_counts / leaf_total
        else:
            leaf_distribution = np.zeros_like(leaf_counts)

        print("Leaf class distribution:")

        for label, count, proportion in zip(
            model.classes_,
            leaf_counts,
            leaf_distribution,
        ):
            print(
                f"  {label}: weighted_count={count:.{decimals}f}, "
                f"proportion={proportion:.{decimals}f}"
            )

    print("=" * 76)


# ---------------------------------------------------------------------
# Explain reproducibly selected test examples
# ---------------------------------------------------------------------

rng = np.random.default_rng(1)

n_examples = min(2, len(X_test_fp))

selected_indices = rng.choice(
    X_test_fp.index.to_numpy(),
    size=n_examples,
    replace=False,
)

for sample_index in selected_indices:
    explain_tree_prediction(
        model=pgfdt,
        row=X_test_fp.loc[sample_index],
        feature_names=X_test_fp.columns,
        true_label=y_test.loc[sample_index],
        show_leaf_distribution=True,
        decimals=3,
    )
    print()
```

```python
# Deliberately pick one correct and one incorrect prediction (rather than relying on
# another random draw), so the explanation report below is guaranteed to show what a
# mistake looks like, not just successes.
test_predictions = pd.Series(
    pgfdt.predict(X_test_fp),
    index=X_test_fp.index,
    name="prediction",
)

correct_indices = test_predictions.index[
    test_predictions == y_test
]

incorrect_indices = test_predictions.index[
    test_predictions != y_test
]

selected_indices = []

if len(correct_indices) > 0:
    selected_indices.append(
        rng.choice(correct_indices.to_numpy())
    )

if len(incorrect_indices) > 0:
    selected_indices.append(
        rng.choice(incorrect_indices.to_numpy())
    )

for sample_index in selected_indices:
    explain_tree_prediction(
        model=pgfdt,
        row=X_test_fp.loc[sample_index],
        feature_names=X_test_fp.columns,
        true_label=y_test.loc[sample_index],
    )
    print()
```

## Summary

- A decision tree restricted to 14 physics-guided Raman fingerprints achieves a grouped cross-validated balanced accuracy of 0.919, outperforming both an unrestricted fixed-width tiling baseline (0.874) and the mean performance of 5,000 equal-budget random window layouts (0.875), while using only a small, interpretable feature set.

- The placement of the fingerprint windows carries meaningful chemical information. The physics-guided dictionary outperformed 88.1% of random window layouts and provided a 4.4 percentage-point improvement over the average random placement. Although this advantage did not reach statistical significance (p=0.119), an unrestricted search independently selected the principal Raman regions associated with cis/trans C=C stretching, ordered C–C skeletal vibrations, and CH stretching, supporting the physical relevance of the proposed fingerprint dictionary.

- Every PG-FDT decision is fully interpretable: each prediction is explained by a sequence of chemically meaningful threshold rules on Raman fingerprints rather than arbitrary spectral variables, allowing every classification to be traced back to known molecular vibrations and lipid chemistry.
