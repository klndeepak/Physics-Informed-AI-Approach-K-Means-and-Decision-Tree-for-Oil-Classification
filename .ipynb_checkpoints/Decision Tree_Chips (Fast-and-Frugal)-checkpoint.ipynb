### Read the Excel file and display the head. 

**Column ordering.** Each spectral column header is a Raman shift (wavenumber, cm⁻¹), not a wavelength. Column order in the source CSV is incidental to how the file was exported and is not guaranteed to carry any meaning, so spectral columns are sorted ascending by wavenumber explicitly below, independent of the source file's order (matching `raman_analysis/data_loading.py::get_spectral_columns` in `Paper3_code`, the reference pipeline this notebook is validated against). This is not only cosmetic: with ~1866 highly correlated neighboring wavenumber columns and only a few hundred training rows, `DecisionTreeClassifier` frequently finds several candidate splits with identical impurity reduction and breaks the tie by column order, so an arbitrary/undocumented column order can silently change which split is chosen.

```python
# Oils dataset
chips_path = r"Paper3_code\Paper3_code\data\Raman_Chips_Baseline_corrected_smoothed_zscorenorm_filtered_500_1900_2600_3000_data_CS.csv"
df_chips = pd.read_csv(chips_path).drop(columns=['Unnamed: 0'])

# --- FIX vs original notebook ---
# The original notebook did `.iloc[:, 2:].iloc[:, ::-1]` here, blindly reversing
# the spectral columns regardless of their source order. With ~1866 highly
# collinear wavenumber channels, sklearn's tree-builder breaks impurity ties
# by feature order, so a reversal can silently change which splits get
# chosen. Fix: sort spectral columns ascending by wavenumber (matches
# raman_analysis/data_loading.py::get_spectral_columns in Paper3_code) so
# column order is well-defined and matches the script pipeline exactly.
def _is_wavenumber(col):
    try:
        float(col)
        return True
    except ValueError:
        return False

meta_cols = [c for c in df_chips.columns if not _is_wavenumber(c)]
spectral_cols = sorted([c for c in df_chips.columns if _is_wavenumber(c)], key=float)
df_chips = df_chips[meta_cols + spectral_cols]

display(df_chips.head())
```

**Why no baseline shift.** This spectrum is already baseline-corrected, smoothed, and Z-score normalized upstream (each wavenumber column centered and scaled across the dataset - see the source filename). A Z-score's negative values are a normal, physically meaningful part of that normalization (intensity below that column's own mean), not an instrument artifact like dark current - so unlike a raw intensity trace, there is no principled "zero" to shift this data to. We only report where the dataset's minimum sits, without altering the data.

```python
# Spectral data starts from column 2
spectra = df_chips.iloc[:, 2:]

# Most negative value in the entire spectral dataset
min_value = spectra.min().min()

# Find row and column containing the minimum value
row_idx, col_idx = (spectra == min_value).stack().idxmax()

print("Most negative value:", min_value)
print("Most negative value:", min_value)
print("Row index:", row_idx)
print("Wavenumber:", col_idx)

# --- FIX vs original notebook ---
# The original notebook shifted every spectrum by +|min_value| here so the
# global minimum becomes 0. This data is already baseline-corrected, smoothed,
# and Z-SCORE NORMALIZED upstream (see filename) - a negative value here is a
# normal, physically meaningful part of a z-score, not an instrument artifact
# to zero out. The Paper3_code pipeline explicitly skips this shift for this
# dataset (DatasetConfig.apply_baseline_shift=False) for exactly this reason.
# Fix: do NOT apply the shift; leave the z-scored spectrum as-is.
print("Shift skipped (data is already Z-score normalized; negative values are physically meaningful)")
print("Minimum stays:", df_chips.iloc[:, 2:].min().min())
```

**Train/test split.** A single stratified 70/30 split is held out up front and reused for every model below (baseline, pre-pruned, post-pruned), so their test-set numbers are directly comparable to each other. `stratify=y` keeps each oil type's proportion the same in both sets; with a few hundred samples split across 5 classes, a plain random split could otherwise leave a class over- or under-represented in the (already small) test set purely by chance.

```python
#Define features (X) and Target (y)
X = df_chips.drop(columns=['Oil Type','Chips Type'])
y= df_chips['Oil Type']
X = X.astype(float)

# Splitting data in train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=1, stratify = y)
print("Shape of Training set : ", X_train.shape)
print("Shape of test set : ", X_test.shape)
print("Percentage of classes in training set:")
print(y_train.value_counts(normalize=True))
print("Percentage of classes in test set:")
print(y_test.value_counts(normalize=True))
```

### Define a helper function to compute Accuracy/Recall/Precision/F1 (weighted) for any classifier.

```python
def model_performance_classification_sklearn(model, predictors, target):
    """
    Function to compute different metrics to check classification model performance

    model: classifier
    predictors: independent variables
    target: dependent variable
    """
    # predicting using the independent variables
    pred = model.predict(predictors)
    acc = accuracy_score(target, pred)
    recall = recall_score(target, pred, average="weighted")
    precision = precision_score(target, pred, average="weighted")
    f1 = f1_score(target, pred, average="weighted")

    # creating a dataframe of metrics
    df_perf = pd.DataFrame({"Accuracy": acc,"Recall": recall,"Precision": precision,"F1": f1,},index=[0],)
    return df_perf
```

### Utility: Confusion matrix to plot a heatmap of the confusion matrix with per-cell percentage annotation.

```python
def confusion_matrix_sklearn(model, predictors, target):
    """
    To plot the confusion_matrix with percentages

    model: classifier
    predictors: independent variables
    target: dependent variable
    """
    y_pred = model.predict(predictors)
    # 5 oil classes -> a 5x5 confusion matrix
    cm = confusion_matrix(target, y_pred)
    labels = np.asarray(        [
            ["{0:0.0f}".format(item) + "\n{0:.2%}".format(item / cm.flatten().sum())]
            for item in cm.flatten()
        ]
    ).reshape(5, 5)

    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=labels, fmt="")
    plt.ylabel("True label")
    plt.xlabel("Predicted label")
```

### Prepruning

### Pre‑pruning: small grid search: Search over `max_depth`, `max_leaf_nodes`, `min_samples_split`; keep the model with the **best test recall** and smallest train–test gap.

**Pre-pruning hyperparameters.** `max_depth` caps how many splits deep the tree can grow, `max_leaf_nodes` caps the total number of terminal nodes, and `min_samples_split` requires a node to hold at least that many training samples before it is allowed to split further - all three restrict tree growth *during* fitting, as opposed to the cost-complexity post-pruning further below, which grows a full tree first and prunes it back afterward. `class_weight="balanced"` reweights each training sample by `n_samples / (n_classes * n_samples_of_its_class)`, so a rarer class counts for more in the split criterion; it is applied here (and in post-pruning) for robustness to any class imbalance the train/test split introduces. Hyperparameters are chosen by 5-fold cross-validated weighted recall on the training set only - the test set is not touched until the winning model is evaluated once, at the end.

```python
# Define the parameters of the tree to iterate over
max_depth_values = np.arange(2, 7, 2)
max_leaf_nodes_values = [50, 75, 150, 250]
min_samples_split_values = [10, 30, 50, 70]
cv = 5
verbose = 1
param_grid = {
"max_depth": list(max_depth_values),
"max_leaf_nodes": list(max_leaf_nodes_values),
"min_samples_split": list(min_samples_split_values),
}
search = GridSearchCV(
DecisionTreeClassifier(class_weight="balanced", random_state=1),
param_grid=param_grid,
scoring="recall_weighted",cv= StratifiedKFold(n_splits=cv, shuffle=True, random_state= 1),refit=True,
verbose=verbose)
search.fit(X_train, y_train)

columns = ["params", "mean_test_score", "std_test_score", "rank_test_score"]
cv_results = (pd.DataFrame(search.cv_results_)[columns].sort_values("rank_test_score").reset_index(drop=True)
)
```

**Reading feature importance.** scikit-learn's Gini importance for a wavenumber sums the (sample-weighted) impurity decrease at every split in the tree that used it, then normalizes so all importances sum to 1. It reflects how much a wavenumber helped separate classes *within this specific tree*, not a model-free measure of how discriminative that Raman band is in isolation: because neighboring wavenumbers are highly correlated, the tree only needs to pick one of several near-equivalent candidates at a split, so an unused neighbor can be just as chemically informative as a used one that scores higher here. Only wavenumbers actually used by a split get a nonzero score, which is why the plot below only shows that nonzero subset.

```python
importances = model1.feature_importances_

# Keep only features used by the tree
mask = importances > 0

important_features = np.array(feature_names)[mask]
important_values = importances[mask]

# Sort by importance
idx = np.argsort(important_values)

plt.figure(figsize=(7,4))

plt.barh(
    range(len(idx)),
    important_values[idx]
)

plt.yticks(
    range(len(idx)),
    important_features[idx]
)

plt.xlabel("Relative Importance")
plt.ylabel("Wavenumber (cm⁻¹)")
plt.title("Pre-pruned Decision Tree Feature Importances")

plt.tight_layout()

plt.savefig(
    "decision_tree_feature_importance_oils.jpg",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

### Post-pruning

### Post‑pruning via Cost‑Complexity Path: Compute `ccp_alphas` and impurities to study pruning trajectory.

For a fully grown tree *T*, cost-complexity pruning defines

R_α(T) = R(T) + α·|leaves(T)|

where R(T) is T's total training impurity and |leaves(T)| its number of leaves. α trades fit against size: α = 0 is the fully grown tree, and increasing α makes larger subtrees costlier, until at each of a finite sequence of α values the optimal subtree collapses its weakest-link node. `cost_complexity_pruning_path` returns exactly that sequence of α values and each one's corresponding leaf impurity.

```python
clf = DecisionTreeClassifier(random_state=1, class_weight="balanced")
path = clf.cost_complexity_pruning_path(X_train, y_train)
# sklearn's ccp_alphas are already >= 0 in practice; abs() guards against a
# stray floating-point sign flip on a value that should be exactly 0.
ccp_alphas, impurities = abs(path.ccp_alphas), path.impurities
```

```python
pd.DataFrame(path)
```

### Impurity vs α (training) from the above summary: Heavier pruning (larger α) reduces tree size and increases leaf impurity.


```python
print(model1.get_n_leaves())
print(model1.get_depth())
```

```python
alphas = path.ccp_alphas

for alpha in alphas:
    clf = DecisionTreeClassifier(random_state=1,ccp_alpha=alpha)
    clf.fit(X_train, y_train)

    print(
        f"alpha={alpha:.6f}",
        "depth=", clf.get_depth(),
        "leaves=", clf.get_n_leaves()
    )
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(ccp_alphas[:-1], impurities[:-1], marker="o", drawstyle="steps-post")
ax.set_xlabel("effective alpha")
ax.set_ylabel("total impurity of leaves (Gini)")
ax.set_title("Total Impurity vs effective alpha for training set")
plt.show()
```

### Fit trees across α values: Train a sequence of pruned trees; we’ll compare metrics across α.


```python
clfs = []
for ccp_alpha in ccp_alphas:
    clf = DecisionTreeClassifier(random_state=1, ccp_alpha=ccp_alpha, class_weight="balanced")
    clf.fit(X_train, y_train)
    clfs.append(clf)
print("Number of nodes in the last tree is: {} with ccp_alpha: {}".format(clfs[-1].tree_.node_count, ccp_alphas[-1]))
```

```python
param_grid = {"ccp_alpha": list(ccp_alphas)}
search = GridSearchCV(DecisionTreeClassifier(random_state=1, class_weight='balanced'),param_grid=param_grid,
scoring="recall_weighted",cv=StratifiedKFold(n_splits= 5, shuffle=True, random_state=1),refit=True,
verbose= 1)
search.fit(X_train, y_train)

cv_table = pd.DataFrame({
"ccp_alpha": param_grid["ccp_alpha"],
"mean_cv_recall": search.cv_results_["mean_test_score"],
"std_cv_recall": search.cv_results_["std_test_score"],
}).sort_values("ccp_alpha").reset_index(drop=True)
```

### Node count / Depth vs α: Sanity‑check that pruning controls complexity as expected.

```python
# Drop the last alpha: it always collapses the tree to a single root node
# (uninformative) - see the last line printed in the loop above (leaves=1).
clfs = clfs[:-1]
ccp_alphas = ccp_alphas[:-1]

node_counts = [clf.tree_.node_count for clf in clfs]
depth = [clf.tree_.max_depth for clf in clfs]
fig, ax = plt.subplots(2, 1, figsize=(10, 7))
ax[0].plot(ccp_alphas, node_counts, marker="o", drawstyle="steps-post")
ax[0].set_xlabel("alpha")
ax[0].set_ylabel("number of nodes")
ax[0].set_title("Number of nodes vs alpha")
ax[1].plot(ccp_alphas, depth, marker="o", drawstyle="steps-post")
ax[1].set_xlabel("alpha")
ax[1].set_ylabel("depth of tree")
ax[1].set_title("Depth vs alpha")
fig.tight_layout()
```

### Recall vs α (Train/Test): Pick α giving the best generalization

As α increases, the tree is pruned more aggressively: small α overfits (a large gap between train and test recall), large α underfits (both curves drop and converge). This curve is diagnostic only - the α actually used below is chosen by cross-validation on the training set (previous section), never by reading a value off this test-set curve, since picking a hyperparameter directly from test-set performance would leak the test set into model selection.

```python
recall_train = []
for clf in clfs:
    pred_train = clf.predict(X_train)
    values_train = recall_score(y_train, pred_train, average='weighted')
    recall_train.append(values_train)
```

```python
recall_test = []
for clf in clfs:
    pred_test = clf.predict(X_test)
    values_test = recall_score(y_test, pred_test, average='weighted')
    recall_test.append(values_test)
```

```python
train_scores = [clf.score(X_train, y_train) for clf in clfs]
test_scores = [clf.score(X_test, y_test) for clf in clfs]
```

```python
fig, ax = plt.subplots(figsize=(15, 5))
ax.set_xlabel("alpha")
ax.set_ylabel("Recall")
ax.set_title("Recall vs alpha for training and testing sets")
ax.plot(
    ccp_alphas, recall_train, marker="o", label="train", drawstyle="steps-post",
)
ax.plot(ccp_alphas, recall_test, marker="o", label="test", drawstyle="steps-post")
ax.axvline(model2.ccp_alpha, color="gray", linestyle="--", linewidth=1, label="CV-selected alpha")
ax.legend()
plt.show()
```

```python
def plot_topk_importances(estimator, feature_names, top_k=20, title="Top feature importances"):
    importances = getattr(estimator, 'feature_importances_', None)
    if importances is None:
        raise ValueError("Estimator has no attribute 'feature_importances_'.")

    imp = pd.Series(importances, index=feature_names)
    imp_nz = imp[imp > 0].sort_values(ascending=False)
    if imp_nz.empty:
        print('All importances are zero (common for some pruned trees).')
        return

    top = imp_nz.head(top_k).sort_values()   # ascending for barh
    plt.figure(figsize=(8, max(3, 0.4*len(top))))
    plt.title(f"{title} (top {len(top)})")
    plt.barh(top.index, top.values, color='mediumpurple')
    plt.xlabel('Gini importance (relative)')
    plt.ylabel('Wavenumber (cm⁻¹)')
    plt.tight_layout()
    plt.show()
```

## Fast-and-Frugal Decision Trees: a "frugal AI" baseline

Everything above is a standard CART decision tree (`DecisionTreeClassifier`): even after pruning, a single prediction can require walking down several internal nodes and can draw on a double-digit number of distinct wavenumbers scattered across the pruned tree's branches. This section asks a different question: **how much classification performance is actually lost if we deliberately restrict ourselves to the simplest possible decision-list model** - one comparison per class, at most, and never more than `n_classes - 1` comparisons total for *any* sample?

This idea comes from the **fast-and-frugal heuristics** literature in judgment and decision-making research (Gigerenzer & Goldstein, 1996; Martignon, Katsikopoulos & Woike, 2008), formalized for structured multi-cue classification as **Fast-and-Frugal Trees (FFTs)** (Phillips, Neth, Woike & Gaissmaier, 2017, *Judgment and Decision Making*). An FFT is a decision list with a **fixed exit structure**: at each level, exactly one cue is checked, and one branch of that check is a terminal ("exit") leaf while the other continues to the next cue - unlike a full CART, which can keep branching in both directions indefinitely. The result is a model a person could execute by hand: *"if intensity at wavenumber X exceeds this value, call it oil type A and stop; otherwise check wavenumber Y..."*.

**Why this dataset in particular.** The chips spectra above show the weakest class separability of the datasets in this project - heating and frying appear to erode some of the oil-type signal present in the unheated oils, and the fully-featured trees above top out well short of the near-perfect accuracy seen elsewhere. If that remaining separability is concentrated in a small number of wavenumbers - as the post-pruned tree's feature-importance plot above already hints, since only a handful of its ~1866 candidate features end up with nonzero importance - then a frugal model restricted to just a few cues should be able to recover most of that same signal. That would be a practically useful finding: an instrument that only needs to measure a handful of specific wavenumbers, rather than a full spectrum, could do nearly as well as one that measures everything. The comparison below tests exactly that.

**How the tree is built.** Starting from all classes "active", at each level we consider every class still active and, for each, fit a depth-1 CART (a "decision stump") to find the single Gini-optimal (feature, threshold) split that best separates that one class from everything else among the samples that have reached this level. Each candidate cue's quality is estimated by 5-fold cross-validated *balanced* accuracy (chosen over plain accuracy since some classes are a small minority at deeper levels), so cue selection - like the hyperparameter selection used for the pre- and post-pruned trees above - never looks at the test set. The class/cue combination with the best CV score "exits" at this level: any sample (train or test) satisfying that single condition is immediately predicted as that class and never evaluated again; every other sample proceeds to the next level, where the process repeats among the remaining classes. Once only one class remains, no further cue is needed - it becomes the default prediction for anything that never exited. For a `K`-class problem this produces at most `K-1` cues total, and any individual sample is classified using at most `K-1` comparisons (often fewer, since it can exit at level 1).

**A limitation, by construction.** A sample whose true class is chosen to exit early only gets one chance to be identified: if it doesn't satisfy that class's exit condition, it proceeds to later levels and can only ever be classified as one of the *remaining* classes from then on - so any such error is permanent, unlike a full CART, which can still route a sample down whichever path best matches all of its features jointly, regardless of exit order. This is precisely the frugality trade-off being investigated here.

```python
from sklearn.metrics import balanced_accuracy_score


class FastFrugalTreeClassifier:
    """A greedy, cross-validated Fast-and-Frugal Tree (FFT) for multiclass data.

    At each level, among the classes not yet "exited", finds the single cue -
    a Gini-optimal single-feature, single-threshold split, from a depth-1
    CART stump - that best identifies one specific class vs everything else,
    evaluated by cross-validated balanced accuracy. The class/cue with the
    best CV score exits at this level (samples satisfying the rule are
    predicted as that class and stop); every other sample continues to the
    next level. The last remaining class needs no rule at all - it is the
    default prediction for anything that never exits. A K-class problem
    therefore never needs more than K-1 cues, and any one sample is
    classified using at most K-1 comparisons.
    """

    def __init__(self, cv=5, random_state=1):
        self.cv = cv
        self.random_state = random_state

    def fit(self, X, y):
        X = pd.DataFrame(X).reset_index(drop=True)
        y = pd.Series(y).reset_index(drop=True)
        self.classes_ = np.array(sorted(y.unique()))
        remaining = list(self.classes_)
        active_idx = np.arange(len(y))
        rules = []

        while len(remaining) > 1:
            best = None  # (cv_score, class, feature, threshold, direction)
            Xa = X.iloc[active_idx]
            ya = y.iloc[active_idx].values

            for c in remaining:
                yb = (ya == c).astype(int)
                # need at least 2 samples on each side to cross-validate this cue at all
                if yb.sum() < 2 or (len(yb) - yb.sum()) < 2:
                    continue
                n_splits = max(2, min(self.cv, yb.sum(), len(yb) - yb.sum()))
                skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=self.random_state)
                scores = []
                for tr, va in skf.split(Xa, yb):
                    stump = DecisionTreeClassifier(max_depth=1, random_state=self.random_state)
                    stump.fit(Xa.iloc[tr], yb[tr])
                    scores.append(balanced_accuracy_score(yb[va], stump.predict(Xa.iloc[va])))
                cv_score = float(np.mean(scores))

                # refit on the full active subset to get the actual deployed cue
                stump = DecisionTreeClassifier(max_depth=1, random_state=self.random_state)
                stump.fit(Xa, yb)
                if stump.tree_.feature[0] < 0:
                    continue  # no usable split found (e.g. already pure)
                feat_name = Xa.columns[stump.tree_.feature[0]]
                thresh = stump.tree_.threshold[0]

                # pick whichever side of the split has the higher concentration of class c
                mask_le = Xa[feat_name].values <= thresh
                rate_le = yb[mask_le].mean() if mask_le.sum() else 0.0
                rate_gt = yb[~mask_le].mean() if (~mask_le).sum() else 0.0
                direction = "le" if rate_le >= rate_gt else "gt"

                if best is None or cv_score > best[0]:
                    best = (cv_score, c, feat_name, thresh, direction)

            if best is None:
                break  # nothing usable left; remaining classes fall through to the default

            cv_score, c, feat_name, thresh, direction = best
            rules.append({"feature": feat_name, "threshold": thresh, "direction": direction,
                           "predict": c, "cv_balanced_acc": cv_score})
            remaining.remove(c)

            feat_vals = X.iloc[active_idx][feat_name].values
            exits = (feat_vals <= thresh) if direction == "le" else (feat_vals > thresh)
            active_idx = active_idx[~exits]
            if len(active_idx) == 0:
                break

        self.rules_ = rules
        self.default_class_ = remaining[0] if remaining else self.classes_[-1]
        self.features_used_ = list(dict.fromkeys(r["feature"] for r in rules))
        return self

    def predict(self, X):
        X = pd.DataFrame(X).reset_index(drop=True)
        pred = np.array([self.default_class_] * len(X), dtype=object)
        decided = np.zeros(len(X), dtype=bool)
        for r in self.rules_:
            vals = X[r["feature"]].values
            cond = (vals <= r["threshold"]) if r["direction"] == "le" else (vals > r["threshold"])
            newly = cond & ~decided
            pred[newly] = r["predict"]
            decided |= newly
        return pred

    def explain(self):
        lines = []
        for i, r in enumerate(self.rules_, 1):
            op = "<=" if r["direction"] == "le" else ">"
            lines.append(f"{i}. IF wavenumber {r['feature']} {op} {r['threshold']:.4g}  "
                          f"-> predict {r['predict']}   (cv balanced acc={r['cv_balanced_acc']:.3f})"
                          f"\n   ELSE continue to next cue")
        lines.append(f"{len(self.rules_) + 1}. (no more cues left) -> predict {self.default_class_}  [default]")
        return "\n".join(lines)
```

```python
fft = FastFrugalTreeClassifier(cv=5, random_state=1)
fft.fit(X_train, y_train)

print("Fast-and-Frugal decision list (evaluate top to bottom, stop at the first match):\n")
print(fft.explain())
print(f"\nDistinct wavenumbers used: {len(fft.features_used_)} of {len(feature_names)} available")
```
